```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
print(os.listdir())
```

```python
movies = pd.read_csv(r"C:\Users\vaish\Downloads\Movie_Recommendation\movies.csv")
print(movies.shape)
movies.head()
```

```python
print(movies.info())
```

```python
print(movies.isnull().sum().sort_values(ascending=False).head(10))
```

```python
print("Duplicate rows:", movies.duplicated().sum())
```

```python
movies["homepage"] = movies["homepage"].fillna("")
movies["tagline"] = movies["tagline"].fillna("")
movies["overview"] = movies["overview"].fillna("")
movies["cast"] = movies["cast"].fillna("")
movies["director"] = movies["director"].fillna("")
```

```python
movies["runtime"] = movies["runtime"].fillna(movies["runtime"].median())
```

```python
movies["release_date"] = movies["release_date"].fillna("Unknown")
```

```python
print(movies.isnull().sum().sort_values(ascending=False).head(10))
```

```python
movies["genres"] = movies["genres"].fillna("")
movies["keywords"] = movies["keywords"].fillna("")
```

```python
print(movies.isnull().sum().sort_values(ascending=False).head(10))
```

```python
feature_columns = [
    "genres",
    "keywords",
    "overview",
    "cast",
    "director",
    "tagline"
]

movies["tags"] = movies[feature_columns].agg(" ".join, axis=1)

movies["tags"] = (
    movies["tags"]
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

movies[["title", "genres", "director", "tags"]].head()
```

```python
tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=10000,
    ngram_range=(1, 2)
)

tfidf_matrix = tfidf.fit_transform(movies["tags"])

print("Number of movies:", tfidf_matrix.shape[0])
print("Number of text features:", tfidf_matrix.shape[1])
print("TF-IDF matrix shape:", tfidf_matrix.shape)
```

```python
similarity = cosine_similarity(tfidf_matrix)

print("Similarity matrix shape:", similarity.shape)
```

```python
def recommend_movies(movie_title, top_n=10):
    matches = movies[
        movies["title"].str.casefold()
        == movie_title.strip().casefold()
    ]

    if matches.empty:
        print("Movie not found. Please check the title.")
        return pd.DataFrame()

    movie_index = matches.index[0]

    scores = list(enumerate(similarity[movie_index]))

    scores = sorted(
        scores,
        key=lambda x: x[1],
        reverse=True
    )

    recommended_indices = [
        idx for idx, score in scores
        if idx != movie_index
    ][:top_n]

    result = movies.iloc[recommended_indices][
        ["title", "genres", "director",
         "vote_average", "release_date", "overview"]
    ].copy()

    result["similarity_score"] = [
        similarity[movie_index, idx]
        for idx in recommended_indices
    ]

    return result.reset_index(drop=True)
```

```python
recommend_movies("Avatar", 10)
```

```python
recommend_movies("The Dark Knight", 10)
```

```python
import json

movie_data = []

for i, row in movies.iterrows():
    scores = sorted(
        enumerate(similarity[i]),
        key=lambda x: x[1],
        reverse=True
    )

    recommendations = []

    for j, score in scores:
        if i == j:
            continue

        recommendations.append(
            str(movies.iloc[j]["title"])
        )

        if len(recommendations) == 10:
            break

    movie_data.append({
        "title": str(row["title"]),
        "genres": str(row.get("genres", "")),
        "overview": str(row.get("overview", "")),
        "recommendations": recommendations
    })

with open("movies.json", "w", encoding="utf-8") as file:
    json.dump(movie_data, file, indent=4, ensure_ascii=False)

```
